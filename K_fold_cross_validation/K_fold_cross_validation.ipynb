# K-fold cross validation
## Introduction

### What is K-fold cross validation?
  - A common problem in model training is limited data. If too much data is used for training, it reduces the amount available for validation, which can lead to bias.
  - Conversely, if too much data is set aside for validation, there may not be enough training data to build a model with high generalization capability.
  - The approach: Cross-validation (often called K-fold Cross-validation)

## The first pattern
  - This is the first pattern for K-fold Cross-validation using basic knowledge about loops in Python.

```python
# Package
import numpy as np
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error # Or accuracy_score for classification
```

```python
# 0. Define dummy data and model for demonstration
from sklearn.linear_model import LinearRegression

# Create a simple synthetic dataset
X = np.random.rand(100, 5)
y = np.random.rand(100)

# Define YourModel as a LinearRegression instance
def YourModel():
    return LinearRegression()
```

```python
# 1. Initialize K-Fold
# n_splits: number of folds (usually 5 or 10)
# shuffle=True: shuffle data before splitting to avoid bias
kf = KFold(n_splits=5, shuffle=True, random_state=42)

fold_scores = []

# 2. Training loop through each Fold
for fold, (train_idx, val_idx) in enumerate(kf.split(X)):

    # Get data for current fold
    # Note: X, y are assumed to be Numpy Arrays. If using Pandas DataFrame, use X.iloc[train_idx]
    X_train, X_val = X[train_idx], X[val_idx]
    y_train, y_val = y[train_idx], y[val_idx]

    # Initialize a new model for each fold
    model = YourModel()

    # Train model
    model.fit(X_train, y_train)

    # Predict on Validation set
    preds = model.predict(X_val)

    # Evaluate and save results
    score = mean_squared_error(y_val, preds)
    fold_scores.append(score)

    print(f"Fold {fold+1} Score: {score:.4f}")

# 3. Aggregate average result and standard deviation
mean_score = np.mean(fold_scores)
std_score = np.std(fold_scores)
print(f"\nMean CV Score: {mean_score:.4f} (± {std_score:.4f})")
```

```python
# 1. Initialize K-Fold
# n_splits: number of folds (usually 5 or 10)
# shuffle=True: shuffle data before splitting to avoid bias
kf = KFold(n_splits=5, shuffle=True, random_state=42)

fold_scores = []

# 2. Training loop through each Fold
for fold, (train_idx, val_idx) in enumerate(kf.split(X)):

    # Get data for current fold
    # Note: X, y are assumed to be Numpy Arrays. If using Pandas DataFrame, use X.iloc[train_idx]
    X_train, X_val = X[train_idx], X[val_idx]
    y_train, y_val = y[train_idx], y[val_idx]

    # Initialize a new model for each fold
    model = YourModel()

    # Train model
    model.fit(X_train, y_train)

    # Predict on Validation set
    preds = model.predict(X_val)

    # Evaluate and save results
    score = mean_squared_error(y_val, preds)
    fold_scores.append(score)

    print(f"Fold {fold+1} Score: {score:.4f}")

# 3. Aggregate average result and standard deviation
mean_score = np.mean(fold_scores)
std_score = np.std(fold_scores)
print(f"\nMean CV Score: {mean_score:.4f} (± {std_score:.4f})")
```

## The second pattern
  - This one will use `cross_val_score` in Scikit-learn

```python
from sklearn.model_selection import cross_val_score, KFold

# Initialize model and splitter
model = YourModel()
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Run K-Fold and automatically return an array of scores for the folds
# The scoring parameter takes a string name of the evaluation function (e.g., 'accuracy', 'neg_mean_squared_error', 'roc_auc')
scores = cross_val_score(model, X, y, cv=kf, scoring='neg_mean_squared_error')

# Since scikit-learn maximizes scores, Loss functions are converted to negative values, so we need to flip the sign
mse_scores = -scores

print(f"All Fold Scores: {mse_scores}")
print(f"Mean CV Score: {np.mean(mse_scores):.4f} (± {np.std(mse_scores):.4f})")
```
