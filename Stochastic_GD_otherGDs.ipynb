## Nhắc lại về Gradient Descent
Quy tắc cập nhật trọng số của Gradient Descent là
$$
\theta = \theta - \eta \nabla \mathcal{L}(\mathbf{\theta})
$$
Trong đó, $\eta$ là hệ số học (learning rate) và $\mathcal{L}(\mathbf{\theta})$ là hàm mất mát (loss function).

Giả sử tập dữ liệu huấn luyện có $N$ điểm dữ liệu $\mathbf{x_i} \in \mathcal{R}^d$, ta định nghĩa bài toán Linear Regression như sau:

$$
y_i = \mathbf{x_i}^T\mathbf{w} + w_0 = \bar{\mathbf{x_i}}^T\bar{\mathbf{w}}
$$

Trong đó,
$$
\bar{\mathbf{x_i}} = \begin{bmatrix}1 \\ x_1 \\ \dots \\ x_d\end{bmatrix}
$$

$$\bar{\mathbf{w}} = \begin{bmatrix}w_0 \\ w_1 \\ \dots \\ w_d\end{bmatrix}$$

Vậy Loss fucntion của bài toán sẽ như sau,

$$
\mathcal{L}(\mathbf{w}) = \frac{1}{2N}\sum_{i=1}^N(y_i - \bar{\mathbf{x_i}}^T \bar{\mathbf{w}})^2 = \frac{1}{2N} \left\|y - \bar{X}\bar{\mathbf{w}}\right\|_2^2
$$

Trong đó,
$$
\bar{X} = \begin{bmatrix}\bar{\mathbf{x_1}}^T \\ \bar{\mathbf{x_2}}^T \\ \dots \\ \bar{\mathbf{x_N}}^T\end{bmatrix}
$$

Quy tắc cập nhật trọng số sẽ là
$$
\bar{\mathbf{w}} = \bar{\mathbf{w}} - \eta \nabla \mathcal{L}(\bar{\mathbf{w}})
$$

## Batch Gradient Descent
Có thể thấy từ đầu đến giờ, ta sử dụng toàn bộ tập huấn luyện để huấn luyện cho mô hình, bằng chứng là ta gộp toàn bộ các điểm dữ liệu vào thành một ma trận duy nhất $\bar{X}$ và sau đó để Gradient Descent xử lý trên ma trận này.

Kỹ thuật sử dụng Gradient Descent duyệt hết toàn bộ các điểm dữ liệu trong một lần được gọi là Batch Gradient Descent (BGD)

Có thể thấy để huấn luyện được mô hình với kỹ thuật BGD thì buộc phải sử dụng toàn bộ tập dữ liệu huấn luyện. Nhưng nếu cơ sở dữ liệu liên tục được cập nhật, mỗi lần thêm một vài điểm dữ liệu mới (người ta gọi đây là Online Learning) thì thuật toán phải tính đạo hàm trên toàn bộ tập dữ liệu một lần nữa, do vậy thì tốn thêm tài nguyên và thời gian tính toán.

Từ đây ta có thể các biến thể để phù hợp với bài toán Online Learning này hơn.

## Stochastic Gradient Descent
Thay vì phải cập nhật trọng số dựa trên đạo hàm của hàm mất mát trên toàn bộ tập dữ liệu, thì thuật toán này sử dụng đạo hàm trên từng điểm dữ liệu $\mathbf{x_i}$ rồi mới cập nhật trọng số trên chính điểm dữ liệu đó.

Sau khi quá trình huấn luyện hoàn tất. Sản phẩm được đem ra sử dụng và có thêm các điểm dữ liệu mới xuất hiện, thuật toán sẽ chỉ cập nhật trọng số dựa trên các điểm dữ liệu mới đó, không cần phải gộp toàn bộ lại rồi mới cập nhật trọng số. Kỹ thuật này còn được gọi là Stochastic Gradient Descent (SGD)

Cú một lần duyệt qua tất cả các điểm dữ liệu trong cơ sở dữ liệu thì ta gọi là một epoch.

Với GD thông thường thì ứng với 1 epoch là một lần cập nhật $\theta$, còn với SGD thì 1 epoch có tới $N$ lần cập nhật $\theta$

Dễ thấy cập nhật trọng số dựa trên từng điểm một thế này sẽ làm tiêu tốn nhiều thời gian tính toán, nhưng SGD lại thường chỉ yêu cầu một lượng epoch nhỏ lúc đầu và các điểm dữ liệu sau đó thì chỉ cần không quá 1 epoch là đã thu được nghiệm tốt, do đó thì SGD thường được sử dụng cho các bài toán yêu cầu dữ liệu thay đổi liên tục (online learning) hay bài toán có lượng cơ sở dữ liệu lớn (Deep learning chẳng hạn).

Vậy thì quy tắc cập nhật trọng số của SGD sẽ như sau:
$$
\theta = \theta - \eta \nabla \mathcal{L}(\mathbf{\theta}; \mathbf{x_i}; y_i)
$$

Với $\mathcal{L}(\mathbf{\theta}; \mathbf{x_i}; y_i)$ là hàm mất mát trên một cặp điểm dữ liệu - nhãn $(\mathbf{x_i}, y_i)$

Ta cũng có thể áp dụng các thuật toán tăng tốc GD dựa trên momentum lên SGD. (sẽ implement nó sau)

Trong Linear regression, thì ta hiểu $\mathbf{\theta} = \mathbf{w}$, Ta có:
$$
\mathcal{L}(\mathbf{w}) = \frac{1}{2}(\mathbf{x_i}\mathbf{w} - y_i)^2
$$


$$
\Rightarrow \nabla_\mathbf{w}\mathcal{L}(\mathbf{w}) = \mathbf{x_i}^T(\mathbf{x_i}\mathbf{w} - y_i)
$$

## SGD with Python

```python
# Collect Data
np.random.seed(42)
N = 5000
X = np.random.rand(N, 5)
X = 2 * X
w = np.random.rand(5, 1)
w_0 = random.random()

# Noise
noise = 0.1 * np.random.randn(N, 1)
y = X @ w + w_0 + noise
y
```

```python
class Stochastic_Gradient_Descent:
    def __init__(self, eta=0.01, epoch=10):
        self.w = None
        self.w_0 = None
        self.loss_history = []
        self.epoch = epoch
        self.eta = eta
        self.nb_iterations = None

    def r2_score(self, X, y):
        y_pred = self.predict(X)
        return 1 - np.sum((y_pred - y)**2) / np.sum((y - np.mean(y))**2)

    def loss(self, y_pred, y_i):
        return (0.5 * (y_pred - y_i) ** 2).item()

    def grad(self, x_i, y_i, y_i_pred):
        return x_i.T * (y_i_pred - y_i)

    def partial_fit(self, X, y):
        N, d = X.shape
        self.nb_iterations = 0

        for num_epoch in range(self.epoch):
            indices = np.random.permutation(N)
            X_shuffled, y_shuffled = X[indices], y[indices]
            epoch_loss = 0.0

            for i in range(N):
                self.nb_iterations += 1
                x_i = X_shuffled[i].reshape(1, -1)
                y_i = y_shuffled[i]

                y_pred = x_i @ self.w + self.w_0
                epoch_loss += self.loss(y_pred, y_i)
                dw = self.grad(x_i, y_i, y_pred)

                self.w -= self.eta * dw
                self.w_0 -= self.eta * (y_pred - y_i).item()

            avg_loss = epoch_loss / N
            self.loss_history.append(avg_loss)

            if avg_loss < 1e-4:
                break
            if len(self.loss_history) > 1 and abs(self.loss_history[-1] - self.loss_history[-2]) < 1e-9:
                break

        return self.loss_history, self.w, self.w_0, self.nb_iterations

    def fit(self, X, y):
        N, d = X.shape
        self.w_0 = 0.0
        self.w = np.zeros((d, 1))
        self.nb_iterations = 0

        for num_epoch in range(self.epoch):
            indices = np.random.permutation(N)
            X_shuffled, y_shuffled = X[indices], y[indices]
            epoch_loss = 0.0

            for i in range(N):
                self.nb_iterations += 1
                x_i = X_shuffled[i].reshape(1, -1)
                y_i = y_shuffled[i]

                y_pred = x_i @ self.w + self.w_0
                epoch_loss += self.loss(y_pred, y_i)
                dw = self.grad(x_i, y_i, y_pred)

                self.w -= self.eta * dw
                self.w_0 -= self.eta * (y_pred - y_i).item()

            avg_loss = epoch_loss / N
            self.loss_history.append(avg_loss)

            if avg_loss < 1e-4:
                break
            if len(self.loss_history) > 1 and abs(self.loss_history[-1] - self.loss_history[-2]) < 1e-9:
                break

        return self.loss_history, self.w, self.w_0, self.nb_iterations

    def predict(self, X):
        return X @ self.w + self.w_0
```

```python
model = Stochastic_Gradient_Descent()
loss_history, w_learned, w_0_learned, nb_iterations = model.fit(X, y)
print(nb_iterations, model.r2_score(X, y))
```

#### Gradient Descent
Đầu tiên, ta đi qua phần code tay cho thuật toán này, sau đó mới đi thực hiện thí nghiệm trên nó

```python
class LinearRegressionScratch:
  def __init__(self, learning_rate=0.01, epoch=5000):
    self.lr = learning_rate
    self.epoch = epoch
    self.weights = None
    self.bias = None
    self.loss_history = []
    self.iters = None

  def r2_score(self, X, y):
    y_pred = self.predict(X)
    return 1 - np.sum((y_pred - y)**2) / np.sum((y - np.mean(y))**2)

  def fit(self, X, y):
    n_samples, n_features = X.shape

    self.weights = np.zeros((n_features, 1))
    self.bias = 0
    self.iters = 0

    for epoch in range(self.epoch):
      self.iters += 1
      y_pred = np.dot(X, self.weights) + self.bias

      Loss = np.mean((y_pred - y) ** 2)
      self.loss_history.append(Loss)

      dw = (2 / n_samples) * np.dot(X.T, (y_pred - y))
      db = (2 / n_samples) * np.sum(y_pred - y)

      if(Loss < 1e-4): break
      if len(self.loss_history) > 1 and abs(self.loss_history[-1] - self.loss_history[-2]) < 1e-9: break

      self.weights -= self.lr *dw
      self.bias -= self.lr * db

    return self.weights, self.bias, self.loss_history

  def predict(self, X):
      return np.dot(X, self.weights) + self.bias
```

```python
# Old Dataset
reg_batch_old = LinearRegressionScratch()
start_GD_with_old_dataset = time.time()
weighs, bias, loss_history = reg_batch_old.fit(X_old, y_old)
end_GD_with_old_dataset = time.time()


print('Learning for the first time with GD with old dataset')
print(f'Execution time with GD using old dataset: {end_GD_with_old_dataset - start_GD_with_old_dataset}s')
print(f"Epoch: {reg_batch_old.iters}, R2 Score: {reg_batch_old.r2_score(X_old, y_old):.4f}")
print(f'Weights {reg_batch_old.weights.T}')
print(f'Bias {reg_batch_old.bias}')
```

```python
# new dataset
X_combined = np.vstack((X_old, X_new))
y_combined = np.concatenate((y_old, y_new))

reg_batch = LinearRegressionScratch()

start_batch_time = time.time()
weights_new, bias_new, loss_history_new = reg_batch.fit(X_combined, y_combined)
end_batch_time = time.time()

print('Learning with GD using new dataset')
print(f'Execution time with GD using new dataset (equals update 50 records): {end_batch_time - start_batch_time}s')
print(f"Epoch: {reg_batch.iters}, R2 Score: {reg_batch.r2_score(X_combined, y_combined):.4f}")
print(f'Weights {reg_batch.weights.T}')
print(f'Bias {reg_batch.bias}')
```
